```python
# 1. ADIM: Veriyi Senin Yönteminle İndir
!pip install bertopic sentence-transformers transformers nltk kaggle
import os
import pandas as pd
import nltk
import re

# Senin API bilgilerini kullanıyoruz
os.environ['KAGGLE_USERNAME'] = 'serkaneren68'
os.environ['KAGGLE_KEY'] = '56535c63a5e74766c651a0d38c9658be'

import kaggle
print("Veri indiriliyor...")
kaggle.api.dataset_download_files('serkaneren68/hepsiburada-reviews', path='./', unzip=True)
```

```python
# 2. ADIM: Veriyi Oku ve Temizle (Noktalama işaretlerini koruyarak!)
df = pd.read_csv('reviews.csv') # İnen dosya adı reviews.csv oluyor
df = df.dropna(subset=['review_text'])

# Yumuşak temizlik (Duygu analizi için şart)
def clean_text(text):
    if not isinstance(text, str): return ""
    # HTML temizle
    text = re.sub(r'<.*?>', ' ', text)
    # Sadece çok bariz bozuk karakterleri temizle ama .,!? kalsın
    text = " ".join(text.split())
    return text

print("Temizlik yapılıyor...")
df['clean_text'] = df['review_text'].apply(clean_text)
```

```python

# 3. ADIM: Cümlelere Böl (Senin dosyan direkt yorumu alıyordu, biz detay istiyoruz)
nltk.download('punkt_tab')
sentences = []
for text in df['clean_text']: # Hız için şimdilik ilk 10.000 yorum
    splits = nltk.sent_tokenize(text)
    for s in splits:
        if len(s) > 15: # Çok kısa cümleleri at
            sentences.append(s)

print(f"Analiz edilecek cümle sayısı: {len(sentences)}")
```

```python
# 4. ADIM: BERTopic (Senin K-Means yerine bunu kullanıyoruz)
from bertopic import BERTopic

# Türkçe destekli model
topic_model = BERTopic(language="multilingual", verbose=True)
topics, probs = topic_model.fit_transform(sentences)

# Sonuçları Göster
topic_model.get_topic_info().head(10)
```

```python
# Sonuçları Göster
topic_model.get_topic_info().head(20)
```

```python
# Outlier (Topic -1) olanları en yakın oldukları konuya zorla dağıt
new_topics = topic_model.reduce_outliers(sentences, topics, strategy="c-tf-idf")

# Modelin topic listesini güncelle
topic_model.update_topics(sentences, topics=new_topics)

# Yeni dağılıma bak
print(topic_model.get_topic_info().head(15))
```
