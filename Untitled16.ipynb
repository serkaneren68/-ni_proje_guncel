```python
# ======================
# 0) Kurulum
# ======================
!pip -q install transformers==4.44.2 accelerate datasets evaluate scikit-learn torch --upgrade

import os, random, json, math, numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, precision_score, recall_score, accuracy_score
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoConfig, AutoModel, get_linear_schedule_with_warmup

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ======================
# 1) Veri yükleme
# ======================
df = pd.read_csv("absa_sample.csv")
ASPECTS = ["performance", "shipping", "durability"]
TEXT_COL = "sentence"

train_df, val_df = train_test_split(df, test_size=0.2, random_state=SEED)
print(f"Train: {len(train_df)} | Val: {len(val_df)}")

# ======================
# 2) Label dönüşümü
# ======================
map_to_idx = {-1:0, 0:1, 1:2}
idx_to_sent = {0:"negative", 1:"none", 2:"positive"}

def encode_labels(row):
    return [map_to_idx[int(row[a])] for a in ASPECTS]

train_labels = np.stack(train_df.apply(encode_labels, axis=1))
val_labels   = np.stack(val_df.apply(encode_labels, axis=1))

# ======================
# 3) Tokenizer
# ======================
MODEL_NAME = "dbmdz/bert-base-turkish-cased"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
MAX_LEN = 160

# ======================
# 4) Dataset sınıfı (sadece text döner)
# ======================
class ABSADataset(Dataset):
    def __init__(self, df, labels):
        self.texts = df[TEXT_COL].tolist()
        self.labels = labels.astype(np.int64)

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        return {
            "text": self.texts[idx],
            "labels": torch.tensor(self.labels[idx], dtype=torch.long)
        }

train_dataset = ABSADataset(train_df, train_labels)
val_dataset   = ABSADataset(val_df, val_labels)

# ======================
# 5) Collate Function (padding burada yapılır)
# ======================
def collate_fn(batch):
    texts = [b["text"] for b in batch]
    labels = torch.stack([b["labels"] for b in batch])
    enc = tokenizer(texts, padding=True, truncation=True, max_length=MAX_LEN, return_tensors="pt")
    enc["labels"] = labels
    return enc

BATCH_SIZE = 8
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, collate_fn=collate_fn)
val_loader   = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn)

# ======================
# 6) Model
# ======================
class MultiHeadBert(nn.Module):
    def __init__(self, model_name, num_aspects=3, num_classes=3):
        super().__init__()
        self.bert = AutoModel.from_pretrained(model_name)
        hidden = self.bert.config.hidden_size
        self.dropout = nn.Dropout(0.1)
        self.classifiers = nn.ModuleList([nn.Linear(hidden, num_classes) for _ in range(num_aspects)])

    def forward(self, input_ids, attention_mask, labels=None):
        out = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        pooled = self.dropout(out.last_hidden_state[:, 0, :])
        logits_list = [head(pooled) for head in self.classifiers]
        logits = torch.stack(logits_list, dim=1)  # [B, 3, 3]
        loss = None
        if labels is not None:
            ce = nn.CrossEntropyLoss()
            losses = [ce(logits[:, i, :], labels[:, i]) for i in range(logits.shape[1])]
            loss = torch.mean(torch.stack(losses))
        return logits, loss

model = MultiHeadBert(MODEL_NAME).to(device)

# ======================
# 7) Eğitim ayarları
# ======================
EPOCHS = 10
LR = 3e-5
optimizer = torch.optim.AdamW(model.parameters(), lr=LR)
total_steps = len(train_loader) * EPOCHS
scheduler = get_linear_schedule_with_warmup(optimizer, 0, total_steps)

# ======================
# 8) Değerlendirme
# ======================
def logits_to_preds(logits):
    return torch.argmax(logits, dim=-1).cpu().numpy()

def evaluate(model, loader):
    model.eval()
    y_true, y_pred = [], []
    with torch.no_grad():
        for batch in loader:
            batch = {k: v.to(device) for k, v in batch.items()}
            logits, _ = model(batch["input_ids"], batch["attention_mask"])
            preds = logits_to_preds(logits)
            y_true.append(batch["labels"].cpu().numpy())
            y_pred.append(preds)
    y_true = np.concatenate(y_true)
    y_pred = np.concatenate(y_pred)
    scores = {}
    for i, a in enumerate(ASPECTS):
        f1 = f1_score(y_true[:, i], y_pred[:, i], average="macro")
        scores[a+"_f1"] = f1
    scores["mean_f1"] = np.mean(list(scores.values()))
    return scores

# ======================
# 9) Eğitim Döngüsü
# ======================
best_f1 = 0
for epoch in range(1, EPOCHS+1):
    model.train()
    total_loss = 0
    for batch in train_loader:
        batch = {k: v.to(device) for k, v in batch.items()}
        optimizer.zero_grad()
        _, loss = model(batch["input_ids"], batch["attention_mask"], batch["labels"])
        loss.backward()
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()
    val_scores = evaluate(model, val_loader)
    print(f"\nEpoch {epoch} | Train Loss: {total_loss/len(train_loader):.4f}")
    print("Validation:", val_scores)
    if val_scores["mean_f1"] > best_f1:
        best_f1 = val_scores["mean_f1"]
        torch.save(model.state_dict(), "best_model.pt")
        print("✅ En iyi model kaydedildi!")

print("\nEğitim tamamlandı. En iyi ortalama F1:", round(best_f1, 4))

# ======================
# 10) Tahmin (örnek)
# ======================
def predict(text):
    model.eval()
    enc = tokenizer([text], padding=True, truncation=True, max_length=MAX_LEN, return_tensors="pt").to(device)
    with torch.no_grad():
        logits, _ = model(enc["input_ids"], enc["attention_mask"])
        preds = torch.argmax(logits, dim=-1).squeeze(0).cpu().numpy()
    return {ASPECTS[i]: idx_to_sent[int(preds[i])] for i in range(len(ASPECTS))}

for ex in [
    "Telefon mükemmel hızda ama kargo çok geç geldi.",
    "Kargo harikaydı, ürün çok sağlam ama performansı kötü."
]:
    print("\n", ex)
    print("Tahmin:", predict(ex))
```

```python
# ======================
# 0) Kurulum
# ======================
!pip -q install transformers==4.44.2 accelerate datasets evaluate scikit-learn torch --upgrade

import os, random, json, math, numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModel, get_linear_schedule_with_warmup

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ======================
# 1) Veri yükleme
# ======================
df = pd.read_csv("absa_sample.csv")  # birleşmiş 150 verilik dosyan
ASPECTS = ["performance", "shipping", "durability"]
TEXT_COL = "sentence"

# %80 train, %10 val, %10 test
train_df, temp_df = train_test_split(df, test_size=0.2, random_state=SEED)
val_df, test_df   = train_test_split(temp_df, test_size=0.5, random_state=SEED)

print(f"Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")

# ======================
# 2) Label dönüşümü (-1→0, 0→1, 1→2)
# ======================
map_to_idx = {-1:0, 0:1, 1:2}
idx_to_sent = {0:"negative", 1:"none", 2:"positive"}

def encode_labels(row):
    return [map_to_idx[int(row[a])] for a in ASPECTS]

train_labels = np.stack(train_df.apply(encode_labels, axis=1))
val_labels   = np.stack(val_df.apply(encode_labels, axis=1))
test_labels  = np.stack(test_df.apply(encode_labels, axis=1))

# ======================
# 3) Tokenizer
# ======================
MODEL_NAME = "dbmdz/bert-base-turkish-cased"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
MAX_LEN = 160

# ======================
# 4) Dataset sınıfı
# ======================
class ABSADataset(Dataset):
    def __init__(self, df, labels):
        self.texts = df[TEXT_COL].tolist()
        self.labels = labels.astype(np.int64)

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        return {
            "text": self.texts[idx],
            "labels": torch.tensor(self.labels[idx], dtype=torch.long)
        }

train_dataset = ABSADataset(train_df, train_labels)
val_dataset   = ABSADataset(val_df, val_labels)
test_dataset  = ABSADataset(test_df, test_labels)

# ======================
# 5) Collate Function (padding burada yapılır)
# ======================
def collate_fn(batch):
    texts = [b["text"] for b in batch]
    labels = torch.stack([b["labels"] for b in batch])
    enc = tokenizer(texts, padding=True, truncation=True, max_length=MAX_LEN, return_tensors="pt")
    enc["labels"] = labels
    return enc

BATCH_SIZE = 8
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, collate_fn=collate_fn)
val_loader   = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn)
test_loader  = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn)

# ======================
# 6) Model
# ======================
class MultiHeadBert(nn.Module):
    def __init__(self, model_name, num_aspects=3, num_classes=3):
        super().__init__()
        self.bert = AutoModel.from_pretrained(model_name)
        hidden = self.bert.config.hidden_size
        self.dropout = nn.Dropout(0.1)
        self.classifiers = nn.ModuleList([nn.Linear(hidden, num_classes) for _ in range(num_aspects)])

    def forward(self, input_ids, attention_mask, labels=None):
        out = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        pooled = self.dropout(out.last_hidden_state[:, 0, :])
        logits_list = [head(pooled) for head in self.classifiers]
        logits = torch.stack(logits_list, dim=1)  # [B, 3, 3]
        loss = None
        if labels is not None:
            ce = nn.CrossEntropyLoss()
            losses = [ce(logits[:, i, :], labels[:, i]) for i in range(logits.shape[1])]
            loss = torch.mean(torch.stack(losses))
        return logits, loss

model = MultiHeadBert(MODEL_NAME).to(device)

# ======================
# 7) Eğitim ayarları
# ======================
EPOCHS = 150
LR = 3e-5
optimizer = torch.optim.AdamW(model.parameters(), lr=LR)
total_steps = len(train_loader) * EPOCHS
scheduler = get_linear_schedule_with_warmup(optimizer, 0, total_steps)

# ======================
# 8) Değerlendirme fonksiyonu
# ======================
def logits_to_preds(logits):
    return torch.argmax(logits, dim=-1).cpu().numpy()

def evaluate(model, loader):
    model.eval()
    y_true, y_pred = [], []
    with torch.no_grad():
        for batch in loader:
            batch = {k: v.to(device) for k, v in batch.items()}
            logits, _ = model(batch["input_ids"], batch["attention_mask"])
            preds = logits_to_preds(logits)
            y_true.append(batch["labels"].cpu().numpy())
            y_pred.append(preds)
    y_true = np.concatenate(y_true)
    y_pred = np.concatenate(y_pred)
    scores = {}
    for i, a in enumerate(ASPECTS):
        f1 = f1_score(y_true[:, i], y_pred[:, i], average="macro")
        scores[a+"_f1"] = f1
    scores["mean_f1"] = np.mean(list(scores.values()))
    return scores

# ======================
# 9) Early Stopping ile Eğitim
# ======================
best_f1 = 0
patience = 10
no_improve = 0

for epoch in range(1, EPOCHS+1):
    model.train()
    total_loss = 0
    for batch in train_loader:
        batch = {k: v.to(device) for k, v in batch.items()}
        optimizer.zero_grad()
        _, loss = model(batch["input_ids"], batch["attention_mask"], batch["labels"])
        loss.backward()
        optimizer.step()
        scheduler.step()
        total_loss += loss.item()

    val_scores = evaluate(model, val_loader)
    print(f"\nEpoch {epoch} | Train Loss: {total_loss/len(train_loader):.4f}")
    print("Validation:", val_scores)

    mean_f1 = val_scores["mean_f1"]
    if mean_f1 > best_f1:
        best_f1 = mean_f1
        torch.save(model.state_dict(), "best_model.pt")
        print(f"✅ Yeni en iyi model kaydedildi! (F1={best_f1:.4f})")
        no_improve = 0
    else:
        no_improve += 1
        print(f"⚠️ İyileşme yok ({no_improve}/{patience})")
        if no_improve >= patience:
            print("⏹ Early stopping tetiklendi.")
            break

print("\nEğitim tamamlandı. En iyi ortalama F1:", round(best_f1, 4))

# ======================
# 10) Test Seti Değerlendirmesi
# ======================
model.load_state_dict(torch.load("best_model.pt", map_location=device))
test_scores = evaluate(model, test_loader)
print("\n📊 Test Sonuçları:", test_scores)

# ======================
# 11) Tahmin (örnek)
# ======================
def predict(text):
    model.eval()
    enc = tokenizer([text], padding=True, truncation=True, max_length=MAX_LEN, return_tensors="pt").to(device)
    with torch.no_grad():
        logits, _ = model(enc["input_ids"], enc["attention_mask"])
        preds = torch.argmax(logits, dim=-1).squeeze(0).cpu().numpy()
    return {ASPECTS[i]: idx_to_sent[int(preds[i])] for i in range(len(ASPECTS))}

for ex in [
    "Telefon mükemmel hızda ama kargo çok geç geldi.",
    "Kargo harikaydı, ürün çok sağlam ama performansı kötü.",
    "Ürün hızlı geldi, performansı da gayet iyi."
]:
    print("\n", ex)
    print("Tahmin:", predict(ex))
```

```python
import torch.nn.functional as F

def predict_with_confidence(text, top_k=3):
    model.eval()
    enc = tokenizer(
        [text],
        padding=True,
        truncation=True,
        max_length=MAX_LEN,
        return_tensors="pt"
    ).to(device)

    with torch.no_grad():
        logits, _ = model(enc["input_ids"], enc["attention_mask"])
        probs = F.softmax(logits, dim=-1).squeeze(0).cpu().numpy()  # [3 aspects, 3 classes]

    results = {}
    print(f"\n🔎 Input: {text}\n")
    for i, aspect in enumerate(ASPECTS):
        aspect_probs = probs[i]
        pred_idx = int(np.argmax(aspect_probs))
        pred_label = idx_to_sent[pred_idx]
        confidence = aspect_probs[pred_idx]

        # En yüksek 3 sınıfı da göstermek istersen:
        top_indices = np.argsort(aspect_probs)[::-1]
        top_info = [(idx_to_sent[j], float(aspect_probs[j])) for j in top_indices]

        print(f"📍 {aspect.upper()} → {pred_label.upper()} ({confidence*100:.2f}% güven)")
        for lbl, val in top_info:
            print(f"   - {lbl:8s}: {val*100:.2f}%")
        print()

        results[aspect] = {
            "predicted": pred_label,
            "confidence": round(float(confidence), 4),
            "distribution": {lbl: round(float(aspect_probs[j]), 4) for j, lbl in enumerate(idx_to_sent.values())}
        }

    return results
```

```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import normalize
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# ============================
# Veri
# ============================
df = pd.read_csv("absa_sample.csv")
sentences = df["sentence"].astype(str).tolist()
print("✅ Toplam cümle:", len(sentences))

# ============================
# Embedding
# ============================
model = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
embeddings = model.encode(sentences, show_progress_bar=True)
embeddings = normalize(embeddings)

# ============================
# Küme sayısı seçimi
# ============================
scores = {}
for k in range(2, 10):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(embeddings)
    sil = silhouette_score(embeddings, labels)
    scores[k] = sil

best_k = max(scores, key=scores.get)
print(f"\n📊 En iyi küme sayısı (silhouette): {best_k}")

# ============================
# Nihai model
# ============================
kmeans = KMeans(n_clusters=best_k, random_state=42)
df["cluster"] = kmeans.fit_predict(embeddings)

# ============================
# Türkçe stopwords listesi
# ============================
turkish_stopwords = [
    "bir", "ve", "çok", "ama", "de", "da", "bu", "şu", "ile", "gibi", "daha",
    "her", "için", "olan", "hem", "fazla", "çünkü", "ben", "sen", "biz",
    "siz", "onlar", "mı", "mi", "mu", "mü", "diye", "yani", "ise", "veya",
    "ya", "değil", "var", "yok", "ki", "göre", "kadar"
]

# ============================
# TF-IDF (manuel stopwords)
# ============================
vectorizer = TfidfVectorizer(
    max_features=2000,
    stop_words=turkish_stopwords,
    ngram_range=(1, 2)
)
X_tfidf = vectorizer.fit_transform(df["sentence"])
terms = np.array(vectorizer.get_feature_names_out())

cluster_keywords = {}
for i in range(best_k):
    idx = np.where(df["cluster"] == i)[0]
    if len(idx) == 0:
        continue
    mean_tfidf = np.mean(X_tfidf[idx].toarray(), axis=0)
    top_idx = mean_tfidf.argsort()[-15:][::-1]
    cluster_keywords[i] = terms[top_idx]

# ============================
# Küme özetleri
# ============================
for i in range(best_k):
    subset = df[df["cluster"] == i]
    print(f"\n🔹 Küme {i+1}")
    print("Anahtar kelimeler:", ", ".join(cluster_keywords[i][:10]))
    print("Temsil cümle örnekleri:")
    for ex in subset.sample(min(3, len(subset))).sentence:
        print("   •", ex)

# ============================
# Küme boyut grafiği
# ============================
plt.figure(figsize=(6, 4))
df["cluster"].value_counts().sort_index().plot(kind="bar", color="lightblue")
plt.title("Küme Dağılımı (Cümle Sayısı)")
plt.xlabel("Küme No")
plt.ylabel("Adet")
plt.show()

# ============================
# WordCloud
# ============================
fig, axes = plt.subplots(1, best_k, figsize=(6*best_k, 5))
if best_k == 1:
    axes = [axes]

for i in range(best_k):
    text = " ".join(df[df["cluster"] == i]["sentence"])
    wc = WordCloud(width=800, height=400, background_color="white", colormap="viridis").generate(text)
    axes[i].imshow(wc, interpolation="bilinear")
    axes[i].axis("off")
    axes[i].set_title(f"Küme {i+1}", fontsize=16)

plt.tight_layout()
plt.show()
```
